```python
pares = [(2,7),(2,0),(10,10),(4,3)]
```

```python
pares_filtrados = []

for primero, segundo in pares:
    if primero > segundo:
        pares_filtrados.append((primero, segundo))

sumas_pares = []
for primero, segundo in pares_filtrados:
    sumas_pares.append((primero + segundo) % 2 == 0)

print("Parejas filtradas:", pares_filtrados)
print("¿Suma es par?:", sumas_pares)
```

#### 9.- Dada la lista _valores_, convierte cada número a string, pero con las siguientes reglas:
- Si es par: "par:<num>"
- Si es múltiplo de 4: "mult3:<num>"
- Si cumple ambas condiciones, concatenar: "par-mult3:<num>"
#### Devuelve la lista final.

```python
valores = [10,21,33,42,55]
```

```python
divisor_multiplo = 4
resultado = []

for numero in valores:
    es_par = numero % 2 == 0
    es_multiplo = numero % divisor_multiplo == 0

    if es_par and es_multiplo:
        etiqueta = f"par-mult3:{numero}"
    elif es_par:
        etiqueta = f"par:{numero}"
    elif es_multiplo:
        etiqueta = f"mult3:{numero}"
    else:
        etiqueta = str(numero)

    resultado.append(etiqueta)

print(resultado)
```

#### 10.- Dado el dicionario _datos_, pon la primera letra de "nombre" en mayúsculas. Calcula la suma total de "puntos" y guarda el total en una nueva clave "total". Marca un booleano como "aprobado" si el total es mayor a 50. Finalmente, crea un set con todos los nombres de los usuarios aprobados. 

```python
datos = [
    {"id":1,"nombre":"paula","puntos":(13,40,60)},
    {"id":2,"nombre":"pepa","puntos":(5,2)},
    {"id":3,"nombre":"mara","puntos":(10,8,20)}
]
```

```python
nombres_aprobados = set()

for persona in datos:
    persona["nombre"] = persona["nombre"].capitalize()
    persona["total"] = sum(persona["puntos"])
    persona["aprobado"] = persona["total"] > 50

    if persona["aprobado"]:
        nombres_aprobados.add(persona["nombre"])

print("Datos actualizados:", datos)
print("Nombres de aprobados:", nombres_aprobados)
```

#### 12.- Dado el diccionario _estudiantes_, crea un nuevo diccionario con la siguiente estructura:
- La clave debe ser el nombre en mayúsculas
- El valor debe ser otro diccionario con:
    - "promedio" como promedio de notas
    - "aprobado" como True si el promedio es >= 20, de lo contrario es False
    - "max_min" como tupla con (nota_max,nota_min)

```python
estudiantes = {
    "s1" : {"nombre":"anastasia","notas":[13,40,60]},
    "s2" : {"nombre":"paula","notas":[5,2]},
    "s3" : {"nombre":"susana","notas":[10,8,20]},
    "s4" : {"nombre":"ainara","notas":[33,17,]}
}
```

```python
resumen_estudiantes = {}

for estudiante in estudiantes.values():
    nombre = estudiante["nombre"].upper()
    notas = estudiante["notas"]
    promedio = sum(notas) / len(notas)

    resumen_estudiantes[nombre] = {
        "promedio": promedio,
        "aprobado": promedio >= 20,
        "max_min": (max(notas), min(notas))
    }

for nombre, resumen in resumen_estudiantes.items():
    print(nombre, resumen)
```

#### 13.- Crea un archivo llamado _notas_alumnos.txt_ con este contenido:
#### Ana,10,20,30
#### Luis,5,15
#### Marta,100,50,25
#### Convierte cada línea en un diccionario con "nombre" capitalizado, "promedio" y "aprobado" si promedio >= 20.
#### Guarda la lista final en un archivo resultado.json

```python
import json

contenido = "Ana,10,20,30\nLuis,5,15\nMarta,100,50,25\n"
with open("notas_alumnos.txt", "w", encoding="utf-8") as archivo:
    archivo.write(contenido)

resultado = []
with open("notas_alumnos.txt", "r", encoding="utf-8") as archivo:
    for linea in archivo:
        campos = linea.strip().split(",")

        nombre = campos[0].capitalize()
        notas = [int(nota) for nota in campos[1:]]
        promedio = sum(notas) / len(notas)

        resultado.append({
            "nombre": nombre,
            "promedio": promedio,
            "aprobado": promedio >= 20
        })
with open("resultado.json", "w", encoding="utf-8") as archivo:
    json.dump(resultado, archivo, ensure_ascii=False, indent=4)

print(json.dumps(resultado, ensure_ascii=False, indent=4))
```

#### 14.- Crea el archivo _usuarios.json_con el contenido del diccionario _datos_. Abre el archivo de nuevo y convierte todos los nombres a mayúsculas, luego crea un archivo _usuarios.csv_ con formato:
- id,nombre,roles
- 1,ANA,user|admin
- 2,LUIS,user
- 3,MARTA,moderator|user

```python
import json
import csv

datos_usuarios = [
    {"id": 1, "nombre": "ana", "roles": ["user", "admin"]},
    {"id": 2, "nombre": "luis", "roles": ["user"]},
    {"id": 3, "nombre": "marta", "roles": ["moderator", "user"]}
]
with open("usuarios.json", "w", encoding="utf-8") as archivo:
    json.dump(datos_usuarios, archivo, ensure_ascii=False, indent=4)
with open("usuarios.json", "r", encoding="utf-8") as archivo:
    usuarios_leidos = json.load(archivo)
for usuario in usuarios_leidos:
    usuario["nombre"] = usuario["nombre"].upper()
with open("usuarios.csv", "w", newline="", encoding="utf-8") as archivo:
    escritor = csv.DictWriter(
        archivo,
        fieldnames=["id", "nombre", "roles"]
    )

    escritor.writeheader()

    for usuario in usuarios_leidos:
        escritor.writerow({
            "id": usuario["id"],
            "nombre": usuario["nombre"],
            "roles": "|".join(usuario["roles"])
        })
with open("usuarios.csv", "r", encoding="utf-8") as archivo:
    print(archivo.read())
```

#### 15.- Con el archivo _fragmento-libro.txt_, ábrelo, cuenta la frecuencia de cada palabra (ignorando mayúsculas/minúsculas y signos). Devuelve un archivo llamado _frecuencias.json_ con las 5 palabras más comunes. 

```python
import json
import re

with open("fragmento-libro.txt", "r", encoding="utf-8") as archivo:
    texto = archivo.read()

palabras = re.findall(r"[^\W\d_]+", texto.lower())

frecuencias = {}

for palabra in palabras:
    frecuencias[palabra] = frecuencias.get(palabra, 0) + 1

palabras_ordenadas = sorted(
    frecuencias.items(),
    key=lambda elemento: elemento[1],
    reverse=True
)

cinco_mas_comunes = dict(palabras_ordenadas[:5])

with open("frecuencias.json", "w", encoding="utf-8") as archivo:
    json.dump(cinco_mas_comunes, archivo, ensure_ascii=False, indent=4)

print(cinco_mas_comunes)
```

#### 16.- Crea tres archivos CSV (_ventas1.csv_, _ventas2.csv_, _ventas3.csv_ con la misma información:
- producto,cantidad
- A,10
- B,5
#### Combínalos en un único archivo que se llema _total_ventas.csv_. Suma las cantidades por producto (si el producto se repite).


```python
import csv

archivos_ventas = ["ventas1.csv", "ventas2.csv", "ventas3.csv"]

# Crear los tres archivos.
for nombre_archivo in archivos_ventas:
    with open(nombre_archivo, "w", newline="", encoding="utf-8") as archivo:
        escritor = csv.writer(archivo)

        escritor.writerow(["producto", "cantidad"])
        escritor.writerow(["A", 10])
        escritor.writerow(["B", 5])

# Leerlos y sumar las cantidades por producto.
totales = {}

for nombre_archivo in archivos_ventas:
    with open(nombre_archivo, "r", newline="", encoding="utf-8") as archivo:
        lector = csv.DictReader(archivo)

        for fila in lector:
            producto = fila["producto"]
            cantidad = int(fila["cantidad"])

            totales[producto] = totales.get(producto, 0) + cantidad

# Guardar los totales en un único archivo.
with open("total_ventas.csv", "w", newline="", encoding="utf-8") as archivo:
    escritor = csv.writer(archivo)
    escritor.writerow(["producto", "cantidad"])

    for producto, cantidad in sorted(totales.items()):
        escritor.writerow([producto, cantidad])

with open("total_ventas.csv", "r", encoding="utf-8") as archivo:
    print(archivo.read())
```

#### 17.- Crea un archivo _alumnos.json_ con el contenido siguiente:
- [{"nombre": "ana", "notas": [10, 20, 30]},{"nombre": "luis", "notas": [5, 15]},{"nombre": "marta", "notas": [100, 50, 25]}]
#### Lee el archivo y genera un diccionario donde la clave sea el nombre y el valor tenga el siguiente formato:
- {"promedio":X,"max":Y,"min":Z,"aprobado":True/False}


```python
import json

datos_alumnos = [
    {"nombre": "ana", "notas": [23, 1, 56]},
    {"nombre": "luis", "notas": [2, 15]},
    {"nombre": "marta", "notas": [9, 29, 33]}
]

umbral_aprobado = 20

with open("alumnos.json", "w", encoding="utf-8") as archivo:
    json.dump(datos_alumnos, archivo, ensure_ascii=False, indent=4)

with open("alumnos.json", "r", encoding="utf-8") as archivo:
    alumnos_leidos = json.load(archivo)

resumen_alumnos = {}

for alumno in alumnos_leidos:
    nombre = alumno["nombre"]
    notas = alumno["notas"]
    promedio = sum(notas) / len(notas)

    resumen_alumnos[nombre] = {
        "promedio": promedio,
        "max": max(notas),
        "min": min(notas),
        "aprobado": promedio >= umbral_aprobado
    }

for nombre, resumen in resumen_alumnos.items():
    print(nombre, resumen)
```
